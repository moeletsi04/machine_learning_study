```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
f = "G:\\My Drive\\INTRO TO ML AND DATA SCIENCE\\UNZIP_FOR_NOTEBOOKS_FINAL\\08-Linear-Regression-Models\\Advertising.csv"
df = pd.read_csv(f)
```

```python
df.head()
```

```python
df['total_spend'] = df['TV'] + df['radio'] +df['sales']
df.head()
```

```python
sns.regplot(data=df,x='total_spend',y='sales') # Seaborn used ordinary least squares
plt.show()
```

```python
x =  df['total_spend']
y = df['sales']


#help(np.polyfit)
beta = np.polyfit(x,y,deg=1) # returns minimized coefficients, gradient then intercept

potential_spend =  np.linspace(0,500,100)
predicted_sales = beta[0]*potential_spend + beta[1] # Highest degree then descending

print(beta)

sns.scatterplot(data=df,x='total_spend',y='sales')
plt.plot(potential_spend,predicted_sales,color='r') # long form of sns.regplot()
plt.show()
```

```python
eta = np.polyfit(x,y,deg=3) 

pred_sales = eta[0]*potential_spend ** 3 + eta[1]*potential_spend**2 + eta[2]*potential_spend + eta[3]

sns.scatterplot(data=df,x='total_spend',y='sales')
plt.plot(potential_spend,pred_sales,color='r') # long form of sns.regplot()
plt.show()
```
